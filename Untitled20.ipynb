```python
# BES_Panel[search(BES_Panel,"(Con|Lab|LD|SNP|PC|BrexitParty|Grn)").index]
cols = search(BES_Panel,"like(Con|Lab|LD|SNP|PC|BrexitParty|Grn)W28").index
mask = ~(BES_Panel[cols].replace("Don't know",np.nan).isnull().all(axis=1))
mostLikedW28 = BES_Panel[cols][mask]\
    .apply(lambda x: x.replace("Don't know",np.nan).cat.codes.replace(-1,np.nan).idxmax(),axis=1)
mostLikedW28 = mostLikedW28.apply(lambda x: x.split("like")[1].split("W")[0]).replace(party_replace_dict)
mostLikedW28.value_counts(dropna=False)

mostLikedW28 = pd.concat([pd.Series([np.nan]*len(mask[~mask]), index=mask[~mask].index), mostLikedW28]).sort_index()
```
