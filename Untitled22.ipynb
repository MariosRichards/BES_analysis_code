```python
%%time
# input_variables = BES_reduced_with_na[search(BES_reduced_with_na,"EUIntegrationSelf|immigSelf|redistSelf|econGenProsp|econGenRetro|prPreference|trustMPs|polAttention|expectGoodConductGeneral|^age$|voterIDSupport").index]

# input_variables = BES_reduced_with_na[search(BES_reduced_with_na,"EUIntegrationSelf|immigSelf|redistSelf|econGenProsp|econGenRetro|prPreference|trustMPs|polAttention|expectGoodConductGeneral|^age$|voterIDSupport").index]
input_variables = BES_reduced_with_na[top20]

target =  BES_Panel["generalElectionVote"].astype('object').replace(replace_dict)
if "generalElectionVoteNonVoter" in BES_Panel.columns:
    target.loc[BES_Panel["generalElectionVoteNonVoter"].notnull()]="DNV"


mask = input_variables.notnull().all(axis=1) &target.notnull()
input_variables = input_variables[mask]
scaler = StandardScaler()
input_variables = pd.DataFrame(scaler.fit_transform(input_variables),columns=input_variables.columns)

target=target[mask]


market_dict = {"LAB":"o","CON":"v","REFUK":"^","LD":"p","GP":"*","OTH":"d"}

import umap
## missing value lines dropped!
# pca = PCA(n_components=3)
# decomp = PCA(svd_solver="full")
# decomp = FactorAnalysis(n_components=2)



# transformed_space = decomp.fit(input_variables).transform(input_variables)
# sloooow
transformed_space = umap.UMAP(n_components=2, n_neighbors=25, min_dist=.01, metric='euclidean',
                              random_state=42, transform_seed=42, verbose=False).fit_transform(input_variables, target_metric='categorical',
                                                             y=target.astype('category').cat.codes.astype('category'))

random_state=42
n_clusters = 12

clust_alg = KMeans(n_clusters=n_clusters, random_state=random_state,)
clust_alg.fit(transformed_space)

fig, ax = plt.subplots(figsize=(16,12))
centers = clust_alg.cluster_centers_

def mscatter(x,y,ax=None, m=None, **kw):
    import matplotlib.markers as mmarkers
    if not ax: ax=plt.gca()
    sc = ax.scatter(x,y,**kw)
    if (m is not None) and (len(m)==len(x)):
        paths = []
        for marker in m:
            if isinstance(marker, mmarkers.MarkerStyle):
                marker_obj = marker
            else:
                marker_obj = mmarkers.MarkerStyle(marker)
            path = marker_obj.get_path().transformed(
                        marker_obj.get_transform())
            paths.append(path)
        sc.set_paths(paths)
    return sc

    

mscatter(transformed_space[:,0],transformed_space[:,1], s=10,
         #c=clust_alg.labels_,
         c=target[mask].replace(party_colour_dict).values,
         # m=list(target[mask].replace(marker_dict).values),
         ax=ax)

# Step size of the mesh. Decrease to increase the quality of the VQ.
h = 0.02  # point in the mesh [x_min, x_max]x[y_min, y_max]
# Plot the decision boundary. For that, we will assign a color to each
x_min, x_max = transformed_space[:, 0].min() - 1, transformed_space[:, 0].max() + 1
y_min, y_max = transformed_space[:, 1].min() - 1, transformed_space[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
clust_alg.cluster_centers_ = clust_alg.cluster_centers_.astype(float)

# Obtain labels for each point in mesh. Use last trained model.
Z = clust_alg.predict(np.c_[xx.ravel(), yy.ravel()])

# Put the result into a color plot
Z = Z.reshape(xx.shape)
# plt.figure(1)
# plt.clf()
ax.imshow(
    Z,
    interpolation="nearest",
    extent=(xx.min(), xx.max(), yy.min(), yy.max()),
    cmap=plt.cm.Paired,
    aspect="auto",
    origin="lower",
    alpha=.3,
)
# up to 11
marker_list = ['o','v','^','x','P','*','X','D',4,5,'H','p']

for cluster_no in range(n_clusters):
    ax.scatter(centers[cluster_no, 0], centers[cluster_no, 1], c=plt.cm.Paired.colors[cluster_no], s=50, label=str(cluster_no),
               marker=marker_list[cluster_no])
plt.xlabel("PCA 0");
plt.ylabel("PCA 1");
ax.legend();
```

```python
from sklearn import metrics
from sklearn.cluster import DBSCAN
plt.figure(figsize=(16,12))
db = DBSCAN(eps=0.5, min_samples=25).fit(transformed_space)
labels = db.labels_

# Number of clusters in labels, ignoring noise if present.
n_clusters_ = len(set(labels)) - (1 if -1 in labels else 0)
n_noise_ = list(labels).count(-1)

print("Estimated number of clusters: %d" % n_clusters_)
print("Estimated number of noise points: %d" % n_noise_)

# print(f"Homogeneity: {metrics.homogeneity_score(labels_true, labels):.3f}")
# print(f"Completeness: {metrics.completeness_score(labels_true, labels):.3f}")
# print(f"V-measure: {metrics.v_measure_score(labels_true, labels):.3f}")
# print(f"Adjusted Rand Index: {metrics.adjusted_rand_score(labels_true, labels):.3f}")
# print(
#     "Adjusted Mutual Information:"
#     f" {metrics.adjusted_mutual_info_score(labels_true, labels):.3f}"
# )
print(f"Silhouette Coefficient: {metrics.silhouette_score(transformed_space, labels):.3f}")


unique_labels = set(labels)
core_samples_mask = np.zeros_like(labels, dtype=bool)
core_samples_mask[db.core_sample_indices_] = True

colors = [plt.cm.Spectral(each) for each in np.linspace(0, 1, len(unique_labels))]

colors = [plt.cm.Paired(each) for each in np.linspace(0, 1, len(unique_labels))]

for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = [0, 0, 0, 1]

    class_member_mask = labels == k

    xy = transformed_space[class_member_mask & core_samples_mask]
    plt.plot(
        xy[:, 0],
        xy[:, 1],
        "o",
        markerfacecolor=tuple(col),
        markeredgecolor="k",
        markersize=14,
    )

    xy = transformed_space[class_member_mask & ~core_samples_mask]
    plt.plot(
        xy[:, 0],
        xy[:, 1],
        "o",
        markerfacecolor=tuple(col),
        markeredgecolor="k",
        markersize=6,
    )

plt.title(f"Estimated number of clusters: {n_clusters_}")
plt.show()
```
