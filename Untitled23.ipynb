```python
pan_dataset_values = pd.read_pickle(BES_small_data_files+"pan_dataset_allr_valuesW19.zip",compression='zip')
```

```python

# likeLabW29,likeLabW30

# pan_dataset_values
q=5

df_temp = BES_Panel[["likeLabW29","likeLabW30", "likeConW29","likeConW30", "likeLDW29","likeLDW30", "likeGrnW29","likeGrnW30", "likeBrexitPartyW29","likeBrexitPartyW30",
                     "lr_scaleW30","al_scaleW30"]]
df_temp["index"]=BES_Panel.index
df_temp["Economic Left-Right"] = df_temp["index"]\
    .apply(lambda x: np.nan if x not in pan_dataset_values.index else pan_dataset_values.loc[x,"Economic: Left <-> Right{gaussian-ish}complete_fully"])
df_temp["Economic Left-Right"]  = pd.qcut(df_temp["Economic Left-Right"],q).cat.codes.replace(-1,np.nan)
df_temp["Social Lib-Auth"] = df_temp["index"]\
    .apply(lambda x: np.nan if x not in pan_dataset_values.index else pan_dataset_values.loc[x,"Social: Liberalism <-> Conservatism{gaussian-ish}complete_fully"])
df_temp["Social Lib-Auth"]  = pd.qcut(df_temp["Social Lib-Auth"],q).cat.codes.replace(-1,np.nan)
df_temp = df_temp.replace("Don't know",np.nan).dropna()
for party in ["Lab","Con","LD","Grn","BrexitParty"]:
    df_temp[["like"+party+"W29","like"+party+"W30"]] = df_temp[["like"+party+"W29","like"+party+"W30"]].apply(lambda x: x.cat.codes.replace(-1,np.nan))
    df_temp["like"+party+"Diff"] = df_temp["like"+party+"W30"] - df_temp["like"+party+"W29"]
```

```python
# df_temp.pivot(index = "Economic Left-Right",columns="Social Lib-Auth",values="likeLabDiff")
for party in ["Lab","Con","LD","Grn","BrexitParty"]:
    plt.figure()
    sns.heatmap( df_temp.groupby(["Economic Left-Right","Social Lib-Auth"])["like"+party+"Diff"].mean().unstack().T[::-1], annot=True );
    plt.title(party+" Support Change");
```
