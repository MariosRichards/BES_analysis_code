```python
temp = BES_Panel[["generalElectionVote","ns_sec","ns_sec_analytic","p_past_vote_2024","workingStatus","p_work_stat","age","ageGroup","wt","p_gross_household"]]
```

```python
temp["age"] = temp["age"].astype('float')
temp["retirementAge"] = temp["age"]>=65
temp.loc[temp["age"].isnull(),"retirementAge"] =np.nan
temp["retirementAge"] = temp["retirementAge"].astype('bool')
```

```python
ct = pd.crosstab(temp["ns_sec_analytic"],temp["generalElectionVote"],values=temp["wt"],aggfunc=sum,normalize='index')*100
ct
```

```python
# sns.barplot(ct.melt())
# plt.xticks(rotation=90);

ct_stack = ct.reset_index().melt(id_vars = ['ns_sec_analytic'])

# sns.barplot( data = ct.reset_index().melt(id_vars = ['ns_sec_analytic']),x='ns_sec_analytic',y='value',hue='generalElectionVote' )

ct_stack.plot(kind='bar', 
                    stacked=True, 
                    colormap='tab10', 
                    figsize=(10, 6))
```

```python
mask = temp["retirementAge"]
temp["generalElectionVote"].value_counts()
```

```python
temp["generalElectionVote"].cat.categories

replace_dict = { 'I would/did not vote':"WNV", 'Conservative':"CON", 'Labour':"LAB", 'Liberal Democrat':"LD",
       'Scottish National Party (SNP)':"SNP", 'Plaid Cymru':"PC", 'Green Party':"GP", 'Other':"OTH",
       'Brexit Party/Reform UK':"REF", "Don't know":"DK" } 

temp["generalElectionVote"] = temp["generalElectionVote"].replace(replace_dict).astype('category')
temp["generalElectionVote"].cat.categories

palette_dict = {'WNV':'k', 'CON':'blue', 'LAB':'red', 'LD':'orange', 'SNP':'yellow', 'PC':'olive', 'GP':'green', 'OTH':'darkgrey', 'REF':'cyan', 'DK':'grey'}
```

```python
tips = sns.load_dataset("tips")

cat_perc = (pd.crosstab(temp["ns_sec_analytic"][mask],
                        temp["generalElectionVote"][mask],
                        values=temp["wt"][mask],aggfunc=sum,normalize='all')*100).sum(axis=1).round().astype('int').to_dict()
temp["ns_sec_analytic_perc"] = temp["ns_sec_analytic"].apply(lambda x: x+" ("+str(cat_perc[x])+"%)")

ax = sns.histplot(
    data=temp[mask],
    y="ns_sec_analytic_perc", hue="generalElectionVote",weights='wt',
    multiple="fill", stat="proportion",hue_order = ['CON', 'REF','DK','OTH','WNV', 'SNP', 'PC',  'LD', 'GP', 'LAB',  ],palette=palette_dict,
    discrete=True, shrink=.8
);
plt.xticks(rotation=90);
sns.move_legend(
    ax, "lower center",
    bbox_to_anchor=(.5, 1), ncol=5, title=None, frameon=False,
)
plt.suptitle("BES Data for Britain split by Vote Intention May 2025: 65+",y=1.05);
plt.ylabel("");
```
