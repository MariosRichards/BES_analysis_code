```python
like_to_party = {'likeCon':"CON", 'likeLab':"LAB", 'likeLD':"LD", 'likeGrn':"GP",  'likeBrexitParty':"REF",}
party_template = {"CON":0, "LAB":0, "LD":0, "GP":0,  "REF":0}
```

```python
def FPTP(df_sub):
    party_dict = party_template.copy()
    winning_party = weighted_value_counts(df_sub[like_parties].idxmax(axis=1),df_sub["wt"],ascending=False)["weighted_sample_size"].index[0]
    party_dict[like_to_party[winning_party]]=1
    return party_dict
```

```python
# for each round select candidate (but could be party) with fewest first-prefs and eliminate
def AV(df_sub):
    winning_parties = like_parties.copy()
    party_dict = party_template.copy()

    while len(winning_parties)>1:
        # could be ambiguous if we have perfect ties - but so is AV?
        losing_party = weighted_value_counts(df_sub[like_parties].idxmax(axis=1),df_sub["wt"],ascending=False)["weighted_sample_size"].index[-1]
        winning_parties = [x for x in winning_parties if x != losing_party]

    # winner takes all  
    party_dict[like_to_party[winning_parties[0]]]=1
    return party_dict

    
```

```python
%%time
# repeats

# store_df = pd.DataFrame() # just number of seats per party???
seats_df = pd.DataFrame(columns = ["repeat","proxy_type","voting_system"]+list(like_to_party.values()))
seats_df_ind = 0
for repeat in range(10):

    for proxy_type in ["likeParty"]:
        
        for voting_system in ["FPTP"]:
            
            store_dict = {} # one dictionary per region
            for reg in df["new_pcon"].unique():

                # and here's your sampling fraction
                df_sub = df[df["new_pcon"]==reg].sample(frac=.66)
                store_dict[reg] = AV(df_sub)
                # get likes, add noise, pick heighest, get *weighted* %, then take top

            # store seat allocation results
            seats_df.loc[seats_df_ind] = [repeat,proxy_type,voting_system]+\
                list(pd.DataFrame( store_dict ).T.sum(axis=0).loc[like_to_party.values()].values)
            seats_df_ind = seats_df_ind+1
## do we want to track both votes and seats? Seats are unquivocal, 'votes' less so in multiround 
        
```
