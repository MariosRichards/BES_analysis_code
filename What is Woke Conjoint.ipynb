```python
# step 1, separate respondent data into a clean data frame
# step 2 build a choice frame with 0/1 for each item if not chosen/chosen
```

```python
RespondentData = WokeData[['CaseID', 'state', 'child_needs_fourie',
       'child_needs_fourie_do', 'climate_medlyn', 'climate_medlyn_do',
       'prejudicelazy_1', 'prejudicelazy_2', 'prejudicelazy_3',
       'prejudicelazy_4', 'prejudicelazy_do', 'religionidpew',
       'religionidpew_do', 'evangelical', 'rid', 'age', 'gender', 'hhi',
       'ethnicity', 'hispanic', 'education', 'political_party', 'region',
       'pid', 'college']].drop_duplicates().set_index("CaseID")
```

```python
RespondentData.isnull().sum().sort_values()
```

```python
# nans only appear under 'evangelical' in large numbers or prejudice_lazy_1/2/3/4, religionidpew, college in small

# no need to fix for xgboost!

# education = ed level, finer breakdown of college y/n

# if the order for prejudice lazy changed, then surely they are not ordinal
```

```python
RespondentDataClean = RespondentData.drop(["rid","prejudicelazy_do",],axis=1)

cat_vars = ["state","region","pid","political_party","hispanic","ethnicity","religionidpew_do","climate_medlyn_do","child_needs_fourie_do",
            "prejudicelazy_1","prejudicelazy_2","prejudicelazy_3","prejudicelazy_4","religionidpew",]
# political_party can be made more legible
RespondentDataClean["political_party"] = RespondentDataClean["pid"].astype(str) +"_" +RespondentDataClean["political_party"].astype(str)
# RespondentDataClean["political_party"].value_counts()



# boolean vars: gender, college, evangelical (with nans)
RespondentDataClean["gender"] = RespondentDataClean["gender"]-1
RespondentDataClean["college"] = RespondentDataClean["college"].cat.codes.replace(-1,np.nan)
RespondentDataClean["evangelical"] = RespondentDataClean["evangelical"].astype('category').cat.codes.replace(-1,np.nan)

# numerical: age, hhi, climate_medlyn, child_needs_fourie, education # fine as they are
temp = pd.get_dummies(RespondentDataClean[cat_vars].astype('category')).astype(float)
RespondentDataClean = RespondentDataClean.drop(cat_vars,axis=1)
RespondentDataClean[temp.columns] = temp
RespondentDataClean = RespondentDataClean.astype(float)
RespondentDataClean.index =RespondentDataClean.index.astype(int)
```

```python
%%time
ind = 0

for row in WokeData[['cjparty', 'cjrace',  'cjsexorient',
       'cjgender', 'cjreligion', 
       'cjcommercialproducts', 'cjpoliticalfigures',
       'cjhistoricalevents',
       'cjprofessions', 'cjcollege', 
       'cjgrouporg',  'cjpoliticalpolicy', ]].values:
    woke_ind = WokeDataCleaned.index[ind]
    # print(ind)
    ind=ind+1
    WokeDataCleaned.loc[woke_ind,row]=1
```
